```python
import re
import nltk
import spacy
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
raw = "The Dogs are RUNNING quickly in the parks!!!"
print(f'Original : {raw}')
```

```python
#lowercasing
step_1 = raw.lower()
print(f"Lowercased : {step_1}\n")
#remove punctuation, !, . @,#
step_2 = re.sub(r"[^a-z\s]","",step_1)
print(f"Remove punctuation : {step_2}\n")
```

```python
stop_words = set(stopwords.words('english'))
print(len(stop_words))
```

```python
step_3 = " ".join(w for w in step_2.split() if w not in stop_words)
print(f"Original : {raw}")
print(f"Stopwords removed : {step_3}")
```

```python
# Stemming
stemmer = PorterStemmer()
stemmed = " ".join(stemmer.stem(w) for w in step_3.split())
print(f"Stemmed : {stemmed}")
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemma = " ".join(lemmatizer.lemmatize(w) for w in step_3.split())
print(f"After Lemmatization : {lemma}")
```

```python
# Sentence tokenazation
text = """"Telangana is a state in southern India on the Deccan Plateau bordering Maharashtra and Chhattisgarh to the north and Andhra Pradesh and Karnataka to the south. It is the eleventh largest and the twelfth most populated state of India. According to the Rigveda's Aitareya Brahmana,
  the region corresponding to Telangana has been inhabited by the Andhras since at least the 9th century BCE with the later Satavahana dynasty, who ruled over the entire Deccan Plateau, establishing trade relations as far as the Roman Empire.
  Subsequent major dynasties include the Vishnukundinas, Eastern Chalukyas, Kakatiyas, the Vijayanagara Empire, and Qutb Shahis,
   followed by the British Raj. Following the Independence of India,
   the Telugu speaking regions of India were gradually regrouped within the state of Andhra Pradesh
   but after decades of protests and agitations, the Telangana movement obtained its bifurcation with the creation, in 2014, of the state of Telangana corresponding to the Telugu speaking regions of the Princely State of Hyderabad."""
sentence_token = nltk.sent_tokenize(text)
for i, sent in enumerate(sentence_token):
  print(f"  Sentence {i+1} : {sent}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
docs = ["the cat sat",
        'the dog sat',
        "the dog and the cat"]

bow_vector = CountVectorizer()
bow_matrix = bow_vector.fit_transform(docs)
bow_df = pd.DataFrame(bow_matrix.toarray(), columns= bow_vector.get_feature_names_out())
bow_df
```

```python
docs = ["the cat sat",
        'the dog sat',
        "the dog and the cat"]
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(docs)
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns= tfidf_vec.get_feature_names_out())
tfidf_df
```
