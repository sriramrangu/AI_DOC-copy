```python
# Inverse of a Matrix
```

```python
# to find inverse of a matrix
#  - matrix should be square matrix
#  - det of matrix should not equal to 0
```

```python
a = np.array([[1,2],[3,4]])
a
```

```python
print(np.linalg.det(a))
```

```python
inverse_a = np.linalg.inv(a)
```

```python
print(inverse_a)
```

```python
print(np.dot(a,inverse_a))
```
